```python
# Clase padre
class Auto:
    def __init__(self, year:int, color:str, carroceria:str, cilindros:int) -> None:
        self.year = year
        self.color = color
        self.carroceria = carroceria
        self.cilindros = cilindros
    
    def hacer_run_run(self):
        print("Run run")
    
    def __str__(self) -> str:
        return f"Este auto es color {self.color}, año {self.year}, es {self.carroceria}, tiene {self.cilindros} cilindros"
```

```python
# Honda hereda de Auto
class Honda(Auto):
    pass
```

```python
civic = Honda(2026, "negro", "sedán", 4)
```

```python
print(civic)
```

```python
civic.hacer_run_run()
```

```python
class Nissan(Auto):
    # Hereda todo tal cual del padre

    # Añade una funcionalidad nueva
    def volar(self):
        print("El auto voló")
```

```python
kicks = Nissan(2025, "blanco", "SUV", 4)
print(kicks)
kicks.hacer_run_run()
```

```python
kicks.volar()
```

```python
civic.volar()
```

```python
# override a un método

class Toyota(Auto):
    # override el init (también recibe caballos (hp))
    def __init__(self, hp:int):
        self.hp = hp

camry = Toyota(126)
```

```python
# Extender a un método

class Toyota(Auto):
    # extender el init (también recibe caballos (hp))
    def __init__(self, year, color, carroceria, cilindros, hp):
        super().__init__(year, color, carroceria, cilindros)
        self.hp = hp

prius = Toyota(2025, "negro", "sedán", 4, 120)
```

```python
# Superclase o clase padre
class Sobreviviente:
    def __init__(self, nombre, salud_max=100) -> None:
        self.nombre = nombre
        self.salud = salud_max
        self.mochila = []

    def buscar_suministros(self):
        print(f"{self.nombre} está buscando suministros en las ruinas...")

    def defenderse(self):
        print(f"{self.nombre} se defiende con un ataque cuerpo a cuerpo básico.")

    def mostrar_estado(self):
        print(f"{self.nombre}\nSalud: {self.salud} | Inventario: {len(self.mochila)} items")

oscar = Sobreviviente("Óscar")
oscar.mostrar_estado()
oscar.defenderse()
```

```python
# Clase hija
class Medico(Sobreviviente):
    # Modificar el init para pedir un dato extra
    def __init__(self, nombre, salud_max=100, botiquines=3) -> None:
        super().__init__(nombre, salud_max)
        # Exclusivo de la clase Medico
        self.botiquines = botiquines

    # Método exclusivo
    def curar_aliado(self, aliado:Sobreviviente):
        if self.botiquines > 0:
            print(f"{self.nombre} usa un botiquín para curar a {aliado.nombre}!")
            self.botiquines -= 1
            aliado.salud = min([100, aliado.salud + 25])
        else:
            print("No tienes botiquines para curar.")
        
shaun = Medico("Shaun", botiquines=5)
miguel = Sobreviviente("Miguel", 80)

shaun.buscar_suministros()
miguel.mostrar_estado()
miguel.salud -= 50
shaun.curar_aliado(miguel)
miguel.mostrar_estado()
```

```python
class Soldado(Sobreviviente):
    def __init__(self, nombre, salud_max=120, municion=30) -> None:
        super().__init__(nombre, salud_max)
        self.municion = municion

    # Override al método del padre
    def defenderse(self):
        if self.municion > 0:
            self.municion -= 1
            print(f"{self.nombre} se defiende utilizando su arma. Munición disponible {self.municion}")
        else:
            print("No tiene munición")
            super().defenderse()

chris = Soldado("Chris Redfield", municion=1)
print("Turno 1".center(50, '-'))
chris.defenderse()

print("Turno 2".center(50, '-'))
chris.defenderse()
```

## POLIMORFISMO
Tercer pilar de la POO

```python
# Crear un escuadrón 
escuadron = [
    Sobreviviente("Óscar"),
    Medico("Shaun", botiquines=1),
    Soldado("Chris")
]
```

```python
print("!UNA HORDA SE ACERCA AL CAMPAMENTO\n")

# A todos darles la misma instrucción
for integrante in escuadron:
    integrante.defenderse() # Polimorfismo
```
