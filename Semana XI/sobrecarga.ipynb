```python
class InventarioJuego:
    def __init__(self, personaje:str) -> None:
        self.personaje = personaje
        self._items = []

    def agregar_item(self, item:str):
        self._items.append(item)
        print(f"{item} se añadió al inventario.")

    # Sobrecargar len()
    def __len__(self):
        return len(self._items)
    
    # Sobrecargar in
    def __contains__(self, item):
        return item in self._items
    
    # Sobrecargar for
    def __iter__(self):
        return iter(self._items)
    
    # Sobrecargar indexing [n]
    def __getitem__(self, key):
        return self._items[key]
```

```python
i1 = InventarioJuego("Miguel")
objetos_a_guardar = ["Arco", "Poción", "Flecha", "Piedra", "Botiquín"]
for objeto in objetos_a_guardar:
    i1.agregar_item(objeto)
len(i1)
```
