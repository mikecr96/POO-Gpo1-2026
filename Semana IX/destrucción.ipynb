```python
# Recolector de basura
class Smartphone:
    def __init__(self, marca, modelo) -> None:
        print('Se ha creado un smartphone.')
        self.marca = marca
        self.modelo = modelo
    # Método destructor
    def __del__(self):
        print('Este smartphone ha muerto.')
```

```python
samsung = Smartphone('Samsung', 'S25 Ultra')
```

```python
del samsung
```
