```python
# Clase abstracta
class MetodosPago(ABC):
    # Método concreto (no es abstracto)
    def informacion(self):
        print("Metodo de pago.")

    # Método abstracto, debe estar vacío
    @abstractmethod
    def pagar(self, monto):
        pass

class PagoEfectivo(MetodosPago):
    def pagar(self, monto):
        print(f"Recibimos {monto} en billetes y monedas. Gracias por tu pago.")

class PagoTarjeta(MetodosPago):
    def pagar(self, monto):
        print(f"Debitamos la cantidad ${monto} de tu tarjeta.")

# objeto1 = MetodosPago() # NO SE PUEDE instanciar a la clase abstracta
# PagoEfectivo.mro()
obj2 = PagoEfectivo()
obj2.informacion()
obj2.pagar(100)

obj3 = PagoTarjeta()
obj3.informacion()
obj3.pagar(100)
```
